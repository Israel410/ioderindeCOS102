```python
import tkinter as tk
from tkinter import messagebox

class DeliveryApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Delivery Quote Tool")
        self.root.geometry("350x300")

        # Inputs
        tk.Label(root, text="Package Weight (kg):").pack(pady=5)
        self.weight_entry = tk.Entry(root)
        self.weight_entry.pack()

        tk.Label(root, text="Location (PAU or Epe):").pack(pady=5)
        self.loc_entry = tk.Entry(root)
        self.loc_entry.pack()

        tk.Button(root, text="Calculate Price", command=self.calculate_fee, bg="blue", fg="white").pack(pady=20)

    def calculate_fee(self):
        try:
            weight = float(self.weight_entry.get())
            location = self.loc_entry.get().strip().lower()
            price = 0

            # Logic from Project Brief
            if location == "pau":
                price = 2000 if weight >= 10 else 1500
            elif location == "epe":
                price = 5000 if weight >= 10 else 4000
            else:
                messagebox.showwarning("Error", "Invalid Location")
                return

            messagebox.showinfo("Quote", f"Total Delivery Fee: N{price}")
        except ValueError:
            messagebox.showerror("Error", "Please enter a valid weight number")

# Run App
if __name__ == "__main__":
    root = tk.Tk()
    app = DeliveryApp(root)
    root.mainloop()
```
