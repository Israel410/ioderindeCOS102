```python
import tkinter as tk
from tkinter import messagebox
import random

class EmployeeSystem:
    def __init__(self, root):
        self.root = root
        self.root.title("Logistics Attendance Portal")
        self.root.geometry("400x300")
        
        # Data from project description
        self.employees = ["Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali"] # Shortened for example
        self.tasks = ["Loading", "Transporting", "Reviewing Orders", "Customer Service"]

        # GUI Elements
        tk.Label(root, text="Enter Employee Name:", font=("Arial", 12)).pack(pady=10)
        self.name_entry = tk.Entry(root, font=("Arial", 12))
        self.name_entry.pack(pady=5)
        tk.Button(root, text="Check In", command=self.check_employee, bg="green", fg="white").pack(pady=20)

    def check_employee(self):
        name = self.name_entry.get()
        # Process 1: Verify Identity
        if name in self.employees:
            self.take_attendance(name)
        else:
            self.refuse_access()

    def take_attendance(self, name):
        # Process 2: Assign Task
        task = random.choice(self.tasks)
        messagebox.showinfo("Success", f"Welcome {name}!\nAttendance marked.\nTask: {task}")

    def refuse_access(self):
        # Process 3: Security Refusal
        messagebox.showerror("Access Denied", "User not found in system.")

# Run App
if __name__ == "__main__":
    root = tk.Tk()
    app = EmployeeSystem(root)
    root.mainloop()
```
