```python
import tkinter as tk
from tkinter import messagebox

# --- OOP Logic ---

class ZenithBank:
    def __init__(self):
        # The project brief says these 3 services are common to most
        self.common = ["Lines of credit", "Investment management", "Insurance"]

class Retail(ZenithBank):
    def get_services(self):
        unique = ["Retirement and education accounts", "Loans and mortgages", "Checking and saving"]
        return self.common + unique

class Global(ZenithBank):
    def get_services(self):
        # Global banking list is entirely unique in your brief
        return ["Multi-currency management", "Foreign currency accounts", 
                "Foreign currency credit cards", "Transborder advisory", "Liquidity management"]

class Commercial(ZenithBank):
    def get_services(self):
        unique = ["Advisory services"]
        return self.common + unique

# --- GUI and Logic Integration ---

class ZenithApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Zenith Bank Division System")
        self.root.geometry("400x250")

        tk.Label(root, text="Enter Employee Name:", font=("Arial", 10)).pack(pady=10)
        self.name_entry = tk.Entry(root)
        self.name_entry.pack(pady=5)

        tk.Button(root, text="Find Services", command=self.match_employee, bg="#E11C22", fg="white").pack(pady=20)

    def match_employee(self):
        # This is where we strictly follow the names from your image
        name = self.name_entry.get().strip().capitalize()
        
        # Mary -> Retail
        if name == "Mary":
            division = Retail()
            div_name = "Retail Banking"
        # Agatha -> Global
        elif name == "Agatha":
            division = Global()
            div_name = "Global Banking"
        # Noel -> Commercial
        elif name == "Noel":
            division = Commercial()
            div_name = "Commercial Banking"
        else:
            messagebox.showerror("Error", f"Employee '{name}' not found in the database.")
            return

        # Polymorphism: Calling the same method name on different class objects
        services = "\n".join([f"• {s}" for s in division.get_services()])
        messagebox.showinfo("Staff Found", f"Name: {name}\nDivision: {div_name}\n\nServices:\n{services}")

if __name__ == "__main__":
    root = tk.Tk()
    app = ZenithApp(root)
    root.mainloop()
```
