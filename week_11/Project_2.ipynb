```python
# PROJECT 1b


from abc import ABC, abstractmethod


# Abstract Base Class
class coup_de_ecriva(ABC):

    @abstractmethod
    def Fan_Page(self):
        pass


# Subclasses implementing Single Inheritance
class FC_Cirok(coup_de_ecriva):

    def Fan_Page(self):
        print('Welcome to "FC Cirok"')


class Madiba_FC(coup_de_ecriva):

    def Fan_Page(self):
        print('Welcome to "Madiba FC"')


class Blue_Jay_FC(coup_de_ecriva):

    def Fan_Page(self):
        print('Welcome to "Blue Jay FC"')


class TSG_Walker(coup_de_ecriva):

    def Fan_Page(self):
        print('Welcome to "TSG Walker"')


# --- Execution Logic ---
def main():
    print("Available Clubs: FC_Cirok, Madiba_FC, Blue_Jay_FC, TSG_Walker")
    user_input = (
        input("Enter the name of the club you support: ").strip().lower()
    )

    # Dictionary mapping user choices to class definitions
    club_mapping = {
        "fc_cirok": FC_Cirok,
        "madiba_fc": Madiba_FC,
        "blue_jay_fc": Blue_Jay_FC,
        "tsg_walker": TSG_Walker,
    }

    # Instantiate the appropriate class and invoke the method
    if user_input in club_mapping:
        club_class = club_mapping[user_input]
        club_instance = club_class()  # Instantiate object
        club_instance.Fan_Page()  # Invoke method
    else:
        print("Invalid club name. Please enter one from the available list.")


if __name__ == "__main__":
    main()
```

```python


#PROJECT 2b 


import tkinter as tk
from abc import ABC, abstractmethod
from tkinter import ttk


# --- Abstract Base Class ---
class External_Vendors(ABC):

    @abstractmethod
    def menu(self):
        """Abstract method to display the vendor menu layout."""
        pass


# --- Subclasses (The Different Vendors) ---
class Cooperative_Cafeteria(External_Vendors):

    def menu(self):
        return [
            ("Jollof Rice and Stew", "₦200"),
            ("White Rice and Stew", "₦200"),
            ("Fried Rice", "₦200"),
            ("Salad", "₦100"),
            ("Plantain", "₦100"),
        ]


class Faith_Hostel_Cafeteria(External_Vendors):

    def menu(self):
        return [
            ("Fried Rice", "₦400"),
            ("White Rice and Stew", "₦400"),
            ("Jollof Rice", "₦400"),
            ("Beans", "₦200"),
            ("Chicken", "₦1000"),
        ]


class Student_Centre_Cafeteria(External_Vendors):

    def menu(self):
        return [
            ("Chicken Fried Rice", "₦800"),
            ("Pomo Sauce", "₦300"),
            ("Spaghetti Jollof", "₦500"),
            ("Amala/Ewedu", "₦500"),
            ("Semo with Efo riro Soup", "₦500"),
        ]


# --- GUI Application Management Class ---
class VendorAppFrame:

    def __init__(self, root):
        self.root = root
        self.root.title("PAU External Food Vendors Management System")
        self.root.geometry("550x450")

        # Mapping dropdown options to their corresponding OOP Subclass Instances
        self.vendor_instances = {
            "Cooperative Cafeteria": Cooperative_Cafeteria(),
            "Faith Hostel Cafeteria": Faith_Hostel_Cafeteria(),
            "Student Centre Cafeteria": Student_Centre_Cafeteria(),
        }

        self._build_ui()

    def _build_ui(self):
        # Header Label
        header = tk.Label(
            self.root,
            text="PAU External Food Vendors",
            font=("Helvetica", 16, "bold"),
            fg="#2C3E50",
            pady=15,
        )
        header.pack()

        # Selection Control Section
        select_frame = tk.Frame(self.root)
        select_frame.pack(pady=5)

        lbl_select = tk.Label(
            select_frame, text="Select Vendor:", font=("Helvetica", 11)
        )
        lbl_select.pack(side=tk.LEFT, padx=5)

        self.vendor_combobox = ttk.Combobox(
            select_frame,
            values=list(self.vendor_instances.keys()),
            state="readonly",
            width=25,
        )
        self.vendor_combobox.pack(side=tk.LEFT, padx=5)
        self.vendor_combobox.set("Select a vendor...")

        # Action Execution Trigger Button
        btn_display = tk.Button(
            self.root,
            text="Display Menu",
            command=self.show_selected_menu,
            bg="#2980B9",
            fg="white",
            font=("Helvetica", 10, "bold"),
            padx=10,
        )
        btn_display.pack(pady=15)

        # Dynamic Content Container Frame
        self.display_container = tk.LabelFrame(
            self.root,
            text=" Live Menu Board ",
            font=("Helvetica", 11, "bold"),
            padx=15,
            pady=15,
        )
        self.display_container.pack(fill="both", expand=True, padx=25, pady=10)

    def show_selected_menu(self):
        # Clear previous records inside container dynamically
        for child in self.display_container.winfo_children():
            child.destroy()

        selected = self.vendor_combobox.get()

        if selected in self.vendor_instances:
            # Instantiate targeted operational object subclass dynamically
            vendor_object = self.vendor_instances[selected]
            # Invoke normal method .menu() explicitly outlined per criteria specifications
            menu_items = vendor_object.menu()

            # Generate Table Title Row
            lbl_item_header = tk.Label(
                self.display_container,
                text="Main Meal",
                font=("Helvetica", 11, "bold"),
            )
            lbl_item_header.grid(row=0, column=0, sticky="w", pady=5)

            lbl_price_header = tk.Label(
                self.display_container,
                text="Price",
                font=("Helvetica", 11, "bold"),
            )
            lbl_price_header.grid(row=0, column=1, sticky="e", pady=5)

            # Auto alignment scaling constraints configurations
            self.display_container.columnconfigure(0, weight=1)
            self.display_container.columnconfigure(1, weight=1)

            # Render matching table item datasets line-by-line
            for index, (meal, price) in enumerate(menu_items, start=1):
                lbl_meal = tk.Label(
                    self.display_container, text=meal, font=("Helvetica", 10)
                )
                lbl_meal.grid(row=index, column=0, sticky="w", pady=3)

                lbl_price = tk.Label(
                    self.display_container, text=price, font=("Helvetica", 10)
                )
                lbl_price.grid(row=index, column=1, sticky="e", pady=3)
        else:
            err_lbl = tk.Label(
                self.display_container,
                text="Please select a valid vendor from the dropdown menu above.",
                fg="#C0392B",
                font=("Helvetica", 10, "italic"),
            )
            err_lbl.pack(pady=20)


# --- Execution Pipeline Activation ---
if __name__ == "__main__":
    window = tk.Tk()
    app = VendorAppFrame(window)
    window.mainloop()
```
