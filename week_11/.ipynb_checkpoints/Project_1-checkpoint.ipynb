```python
import os
import pandas as pd


class StudentClassifier:

    def __init__(self, csv_file_path):
        """Initializes the classifier and loads the SIS data."""
        self.csv_file_path = csv_file_path
        self.df = None
        # Private attribute to store error messages for encapsulated requests (> 25)
        self.__error_message = ""

    def load_data(self):
        """Loads data from the CSV file."""
        if os.path.exists(self.csv_file_path):
            self.df = pd.read_csv(self.csv_file_path)
            print("Successfully loaded SIS data.\n")
        else:
            raise FileNotFoundError(f"The file {self.csv_file_path} was not found.")

    def classify_and_save(self):
        """Filters students into categories based on age rules and saves to CSVs."""
        if self.df is None:
            print("No data loaded to classify.")
            return

        # 1. The Pirates: Greater than 14 and less than 18
        pirates_df = self.df[(self.df["Age"] > 14) & (self.df["Age"] < 18)]
        self._save_and_display(pirates_df, "The_Pirates")

        # 2. The Yankees: Greater than 18 and less than 22
        yankees_df = self.df[(self.df["Age"] > 18) & (self.df["Age"] < 22)]
        self._save_and_display(yankees_df, "The_Yankees")

        # 3. The Bulls: Greater than 22 and less than 25
        bulls_df = self.df[(self.df["Age"] > 22) & (self.df["Age"] < 25)]
        self._save_and_display(bulls_df, "The_Bulls")

        # 4. Age greater than 25: Encapsulate the request / error handling
        invalid_students = self.df[self.df["Age"] > 25]
        if not invalid_students.empty:
            count = len(invalid_students)
            # Setting the private error message attribute
            self.__error_message = (
                f"Error: Found {count} student(s) with an age greater than 25. "
                f"Access denied for classification."
            )
        else:
            self.__error_message = "No age violations (> 25) detected."

    def _save_and_display(self, filtered_df, category_name):
        """Helper method to save to csv and print out results."""
        print(f"=== {category_name.replace('_', ' ')} ===")
        if not filtered_df.empty:
            filename = f"{category_name}.csv"
            filtered_df.to_csv(filename, index=False)
            print(f"Saved successfully to '{filename}'")
            print(filtered_df)
        else:
            print("No students found in this age category.")
        print("-" * 50)

    # Getter method to safely retrieve the encapsulated error message
    def get_error_message(self):
        """Getter method to return the encapsulated error message."""
        return self.__error_message


# --- Execution Block ---
if __name__ == "__main__":
    # Ensure you have 'SIS.csv' in the same folder where you run this script.
    # If using Jupyter, make sure the CSV is in your active workspace.
    classifier = StudentClassifier("SIS.csv")

    try:
        classifier.load_data()
        classifier.classify_and_save()

        print("\n=== Encapsulation Check (Getter Method) ===")
        print(classifier.get_error_message())

    except Exception as e:
        print(f"An error occurred: {e}")
```
